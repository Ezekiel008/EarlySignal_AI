## Imports

```python
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("EarlySignal AI — Notebook 01")
print("Data Understanding & Validation")
```

## Set project paths

```python
PROJECT_ROOT = Path.cwd().parent

DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "synthetic"
    / "earlysignal_monitoring_data.csv"
)

print("Project root:", PROJECT_ROOT)
print("Dataset path:", DATA_PATH)
print("Dataset exists:", DATA_PATH.exists())
```

## Load the dataset

```python
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

## Dataset structure

```python
print("Shape:", df.shape)

print("\nColumn names:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i:02d}. {column}")
```

## Data types

```python
df.info()
```

## Missing values

```python
missing = (
    df.isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame("missing_values")
)

missing["missing_pct"] = (
    missing["missing_values"] / len(df) * 100
).round(2)

missing
```

## Duplicate records

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
print(
    "Duplicate percentage:",
    round(duplicate_rows / len(df) * 100, 2),
    "%"
)
```

## Numerical summary

```python
df.describe(include="number").T
```

## Categorical summary

```python
categorical_cols = df.select_dtypes(
    include=["object", "string", "category"]
).columns

for col in categorical_cols:
    print(f"\n{col}")
    print("-" * len(col))
    print(df[col].value_counts(dropna=False).head(15))
```

## Risk-label distribution

```python
risk_distribution = df["risk_label"].value_counts()
risk_percentage = df["risk_label"].value_counts(
    normalize=True
).mul(100).round(2)

risk_summary = pd.DataFrame({
    "count": risk_distribution,
    "percentage": risk_percentage
})

risk_summary
```

## Anomaly distribution

```python
anomaly_summary = pd.DataFrame({
    "count": df["anomaly_flag"].value_counts(),
    "percentage": df["anomaly_flag"]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
})

anomaly_summary
```

## Basic validation checks

```python
checks = {
    "Monthly target is positive":
        (df["monthly_target"] > 0).all(),

    "Monthly achievement is non-negative":
        (df["monthly_achievement"] >= 0).all(),

    "Achievement rate is non-negative":
        (df["achievement_rate"] >= 0).all(),

    "Activity completion rate is valid":
        df["activity_completion_rate"].between(0, 1).all(),

    "Budget utilisation rate is within plausible range":
        df["budget_utilisation_rate"].between(0, 1.2).all(),

    "Staff availability rate is valid":
        df["staff_availability_rate"].between(0, 1).all(),

    "Reporting delay is non-negative":
        (df["reporting_delay_days"] >= 0).all(),

    "Supply delay is non-negative":
        (df["supply_delay_days"] >= 0).all(),

    "Access constraint score is valid":
        df["access_constraint_score"].between(0, 1).all(),

    "Data quality score is valid":
        df["data_quality_score"].between(0, 1).all(),
}

validation_results = pd.Series(
    checks,
    name="passed"
).to_frame()

validation_results
```

## Early warning preview

```python
high_priority_preview = (
    df[df["risk_label"] == "High"]
    .sort_values(
        ["achievement_rate", "reporting_delay_days"],
        ascending=[True, False]
    )
)
columns_to_show = [
    "reporting_date",
    "programme_name",
    "sector",
    "state",
    "lga",
    "monthly_target",
    "monthly_achievement",
    "achievement_rate",
    "activity_completion_rate",
    "budget_utilisation_rate",
    "reporting_delay_days",
    "risk_label",
    "anomaly_flag",
    "recommended_action"
]

high_priority_preview[columns_to_show].head(20)
```

## Panel Structure Validation output

```python
panel_duplicates = (
    df.groupby(
        [
            "programme_name",
            "state",
            "lga",
            "community",
            "reporting_date"
        ]
    )
    .size()
)

duplicate_unit_dates = (
    panel_duplicates > 1
).sum()

expected_months = (
    df["reporting_date"]
    .nunique()
)

units = (
    df[
        [
            "programme_name",
            "state",
            "lga",
            "community"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)

print("PANEL STRUCTURE VALIDATION")
print("-" * 40)

print(
    "Programme/location/community units:",
    units
)

print(
    "Unique reporting months:",
    expected_months
)

print(
    "Duplicate unit-month combinations:",
    duplicate_unit_dates
)

print(
    "Expected panel rows:",
    units * expected_months
)

print(
    "Actual panel rows:",
    len(df)
)

print(
    "\nPanel structure valid:",
    duplicate_unit_dates == 0
    and len(df) == units * expected_months
)
```
